Problem 9

```python
# Store the total cost and number of traders in variables
total_cost = 1800
traders = 3

# Calculate the cost per trader
cost_per_trader = total_cost / traders

# Print the result clearly
print("Cost per trader:", cost_per_trader, "LRD")
```

Problem 10

```python
# Store the original price and discount rate in variables
original_price = 2500
discount_rate = 0.10

# Calculate the discount amount
discount_amount = original_price * discount_rate

# Calculate the final price after discount
final_price = original_price - discount_amount

# Print both results clearly
print("Discount amount:", discount_amount, "LRD")
print("Final price:", final_price, "LRD")
```
